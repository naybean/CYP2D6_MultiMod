# CYP2D6 Multimodal CNN+GCN — Diplotype-disjoint 10-fold CV

Main-performance evaluation using all `dataset.csv`. Uses the original CNN genotype encoder, original RDKit atom features + GCN drug encoder, and concatenation fusion. Exact diplotypes are group-held-out with the same deterministic folds as the genotype-only notebook.

## 1. Setup

```python
import os, random, warnings
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader, Subset
from sklearn.model_selection import StratifiedGroupKFold
from sklearn.preprocessing import LabelEncoder, label_binarize
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score,
                             roc_auc_score, average_precision_score, confusion_matrix,
                             classification_report, log_loss)
import matplotlib.pyplot as plt

SEED = 42
N_SPLITS = 5
INNER_SPLITS = 5
BATCH_SIZE = 32

def seed_everything(seed=SEED):
    random.seed(seed); np.random.seed(seed); torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)
        torch.backends.cudnn.deterministic = True
        torch.backends.cudnn.benchmark = False
seed_everything()

device = torch.device('cuda:1' if torch.cuda.is_available() else 'cpu')
print('Device:', device)

from torch_geometric.nn import global_mean_pool, GCNConv
from torch_geometric.data import Data, Batch
from rdkit import Chem
```

## 2. Load all 576 samples and define diplotype groups

```python
# data_576.csv 전체를 사용합니다.
DATA_CSV = 'dataset.csv'

data = pd.read_csv(DATA_CSV).copy()
required = {'Variant', 'Class'}
assert required.issubset(data.columns), f'Missing columns: {required - set(data.columns)}'

def canonical_diplotype(v):
    # A; B와 B; A를 같은 diplotype group으로 처리
    alleles = [x.strip() for x in str(v).split(';') if x.strip()]
    return '; '.join(sorted(alleles))

data['DiplotypeGroup'] = data['Variant'].map(canonical_diplotype)
label_encoder = LabelEncoder()
data['ClassEncoded'] = label_encoder.fit_transform(data['Class'])
class_mapping = {i: c for i, c in enumerate(label_encoder.classes_)}
print('Shape:', data.shape)
print('Unique diplotype groups:', data['DiplotypeGroup'].nunique())
print('Unique drugs:', data['Drug'].nunique() if 'Drug' in data else 'NA')
print('Class mapping:', class_mapping)
print(data['Class'].value_counts())
```

## 3. Load CYP2D6 one-hot matrices

```python
ONEHOT_FOLDER = 'onehot_7'
onehot_files = os.listdir(ONEHOT_FOLDER)
onehot_matrices = {}
max_rows = 0
for file in onehot_files:
    if not file.endswith('.csv'): continue
    variant_name = file.split('.csv')[0].replace('_', '.')
    matrix = pd.read_csv(os.path.join(ONEHOT_FOLDER, file), header=None).values
    onehot_matrices[variant_name] = matrix
    max_rows = max(max_rows, matrix.shape[0])
print('Loaded matrices:', len(onehot_matrices), '| max rows:', max_rows)

def pad_matrix(matrix, target_rows=4330, target_cols=7):
    current_rows, current_cols = matrix.shape
    padded = torch.zeros((target_rows, target_cols), dtype=torch.float32)
    padded[:current_rows, :current_cols] = torch.tensor(matrix, dtype=torch.float32)
    return padded
for key in list(onehot_matrices):
    onehot_matrices[key] = pad_matrix(onehot_matrices[key])

def map_variant_to_onehot(variant):
    variants = [x.strip() for x in str(variant).split(';')]
    matrices = []
    for v in variants[:2]:
        if v not in onehot_matrices:
            raise KeyError(f'One-hot matrix not found: {v}')
        matrices.append(onehot_matrices[v])
    while len(matrices) < 2:
        matrices.append(torch.zeros((4330, 7), dtype=torch.float32))
    return torch.vstack(matrices[:2])

X = torch.stack([map_variant_to_onehot(v) for v in data['Variant']])
y = torch.tensor(data['ClassEncoded'].values, dtype=torch.long)
groups = data['DiplotypeGroup'].values
print('X:', X.shape, 'y:', y.shape)
```

## 4. Create the shared 10-fold diplotype-disjoint outer split

```python
# 동일한 outer split을 두 모델에서 재현할 수 있도록 fold assignment를 생성/저장합니다.
outer_cv = StratifiedGroupKFold(n_splits=N_SPLITS, shuffle=True, random_state=SEED)
fold_id = np.full(len(data), -1, dtype=int)

for fold, (outer_train_idx, test_idx) in enumerate(outer_cv.split(np.zeros(len(y)), y.numpy(), groups)):
    train_groups = set(groups[outer_train_idx]); test_groups = set(groups[test_idx])
    overlap = train_groups & test_groups
    assert len(overlap) == 0, f'Fold {fold+1}: diplotype leakage detected: {overlap}'
    fold_id[test_idx] = fold + 1
    counts = np.bincount(y.numpy()[test_idx], minlength=len(label_encoder.classes_))
    print(f'Fold {fold+1}: train={len(outer_train_idx)}, test={len(test_idx)}, '
          f'train groups={len(train_groups)}, test groups={len(test_groups)}, overlap=0, class counts={counts.tolist()}')

assert np.all(fold_id > 0)
split_df = data[['Variant','DiplotypeGroup','Drug','Class']].copy()
split_df['fold'] = fold_id
split_df.to_csv('diplotype_grouped_5fold_assignments.csv', index=False)
print('Saved: diplotype_grouped_5fold_assignments.csv')
```

## 5. Metrics and inner validation split

```python
def compute_metrics(y_true, probs, n_classes=3):
    y_true = np.asarray(y_true); probs = np.asarray(probs)
    pred = probs.argmax(axis=1)
    out = {
        'accuracy': accuracy_score(y_true, pred),
        'precision_macro': precision_score(y_true, pred, average='macro', zero_division=0),
        'recall_macro': recall_score(y_true, pred, average='macro', zero_division=0),
        'f1_macro': f1_score(y_true, pred, average='macro', zero_division=0),
        'f1_weighted': f1_score(y_true, pred, average='weighted', zero_division=0),
        'nll': log_loss(y_true, probs, labels=list(range(n_classes)))
    }
    try: out['auroc_macro_ovr'] = roc_auc_score(y_true, probs, multi_class='ovr', average='macro')
    except ValueError: out['auroc_macro_ovr'] = np.nan
    aps=[]
    for c in range(n_classes):
        binary=(y_true==c).astype(int)
        aps.append(average_precision_score(binary, probs[:,c]) if binary.sum()>0 else np.nan)
    out['aupr_macro_ovr'] = float(np.nanmean(aps))
    return out

def make_inner_split(outer_train_idx, fold):
    # outer-test와 완전히 분리된 validation set. inner split도 diplotype-disjoint.
    y_outer = y.numpy()[outer_train_idx]
    g_outer = groups[outer_train_idx]
    inner_cv = StratifiedGroupKFold(n_splits=INNER_SPLITS, shuffle=True, random_state=SEED + fold)
    inner_train_rel, val_rel = next(inner_cv.split(np.zeros(len(outer_train_idx)), y_outer, g_outer))
    train_idx = outer_train_idx[inner_train_rel]
    val_idx = outer_train_idx[val_rel]
    assert not (set(groups[train_idx]) & set(groups[val_idx]))
    assert not (set(groups[train_idx]) & set(groups[fold_id == fold]))
    return train_idx, val_idx
```

## 6. Multimodal dataset / SMILES graph conversion

```python
class MultimodalDataset(Dataset):
    def __init__(self, X, smiles, y): self.X,self.smiles,self.y=X,smiles,y
    def __len__(self): return len(self.y)
    def __getitem__(self,idx): return self.X[idx].float().unsqueeze(0), smiles_to_graph(self.smiles[idx]), self.y[idx].long()

def one_of_k_encoding(x, allowable_set):
    if x not in allowable_set: x=allowable_set[-1]
    return [int(x==s) for s in allowable_set]

def atom_to_onehot_features(atom):
    return (one_of_k_encoding(atom.GetSymbol(),['H','C','N','O','F','P','S','Cl','Br','I','other']) +
            one_of_k_encoding(atom.GetDegree(),[0,1,2,3,4,5]) +
            one_of_k_encoding(atom.GetExplicitValence(),[0,1,2,3,4,5,6]) +
            one_of_k_encoding(atom.GetImplicitValence(),[0,1,2,3,4,5]) +
            one_of_k_encoding(atom.GetFormalCharge(),[-1,0,1]) +
            one_of_k_encoding(atom.GetHybridization(),[Chem.rdchem.HybridizationType.SP,Chem.rdchem.HybridizationType.SP2,Chem.rdchem.HybridizationType.SP3,'other']) +
            [atom.GetIsAromatic()] +
            one_of_k_encoding(atom.GetChiralTag(),[Chem.rdchem.ChiralType.CHI_UNSPECIFIED,Chem.rdchem.ChiralType.CHI_TETRAHEDRAL_CW,Chem.rdchem.ChiralType.CHI_TETRAHEDRAL_CCW,Chem.rdchem.ChiralType.CHI_OTHER]))

def smiles_to_graph(smiles):
    mol=Chem.MolFromSmiles(smiles)
    if mol is None: raise ValueError(f'Invalid SMILES: {smiles}')
    mol=Chem.AddHs(mol)
    x=torch.tensor([atom_to_onehot_features(a) for a in mol.GetAtoms()],dtype=torch.float32)
    rows=[]; cols=[]
    for b in mol.GetBonds():
        s,e=b.GetBeginAtomIdx(),b.GetEndAtomIdx(); rows += [s,e]; cols += [e,s]
    edge_index=torch.tensor([rows,cols],dtype=torch.long)
    return Data(x=x,edge_index=edge_index)

def collate_fn(batch):
    cnn_inputs,gcn_data,labels=zip(*batch)
    return torch.stack(cnn_inputs), Batch.from_data_list(gcn_data), torch.stack(labels)

smiles=data['isomeric'].astype(str).values
dataset=MultimodalDataset(X,smiles,y)
print('Atom feature dimension:',smiles_to_graph(smiles[0]).x.shape[1])
```

## 7. Original multimodal architecture

```python
class SimpleCNN(nn.Module):
    # 기존 multimodal CNN feature extractor 유지
    def __init__(self):
        super().__init__()
        self.conv1=nn.Conv2d(1,8,kernel_size=(20,7),stride=(1,1)); self.pool1=nn.MaxPool2d((2,1))
        self.conv2=nn.Conv2d(8,16,kernel_size=(20,1),stride=(1,1)); self.pool2=nn.MaxPool2d((2,1))
        self.fc1=nn.Linear(34400,512); self.fc2=nn.Linear(512,128)
    def forward(self,x):
        x=self.pool1(F.relu(self.conv1(x))); x=self.pool2(F.relu(self.conv2(x))); x=x.view(x.size(0),-1)
        return F.relu(self.fc2(F.relu(self.fc1(x))))

class GCNModel(nn.Module):
    def __init__(self,num_node_features=42,hidden_channels=64):
        super().__init__(); self.conv1=GCNConv(num_node_features,hidden_channels); self.conv2=GCNConv(hidden_channels,hidden_channels); self.conv3=GCNConv(hidden_channels,hidden_channels)
        self.bn1=nn.BatchNorm1d(hidden_channels); self.bn2=nn.BatchNorm1d(hidden_channels); self.bn3=nn.BatchNorm1d(hidden_channels); self.fc=nn.Linear(hidden_channels,128)
    def forward(self,data):
        x,e=data.x,data.edge_index; x=F.relu(self.bn1(self.conv1(x,e))); x=F.relu(self.bn2(self.conv2(x,e))); x=self.bn3(self.conv3(x,e)); x=global_mean_pool(x,data.batch); return F.relu(self.fc(x))

class MultimodalModel(nn.Module):
    def __init__(self):
        super().__init__(); self.cnn_model=SimpleCNN(); self.gcn_model=GCNModel(); self.fc1=nn.Linear(256,64); self.fc2=nn.Linear(64,3)
    def forward(self,cnn_input,gcn_data):
        z=torch.cat([self.cnn_model(cnn_input),self.gcn_model(gcn_data)],dim=1)
        return self.fc2(F.relu(self.fc1(z)))
```

## Model training and outer-fold evaluation

```python
# ==========================================
# Multimodal model training and outer-fold evaluation
# ==========================================
EPOCHS = 100
PATIENCE = 30
cnn_lr = 0.0005
gcn_lr = 0.00025
fc_lr = 0.0005

criterion = nn.CrossEntropyLoss()

fold_rows = []
oof_probs = np.zeros((len(data), 3), dtype=float)
oof_pred = np.full(len(data), -1)

for fold in range(1, N_SPLITS + 1):
    print('\n' + '=' * 90)
    print(f'OUTER FOLD {fold}/{N_SPLITS}')
    print('=' * 90)

    test_idx = np.where(fold_id == fold)[0]
    outer_train_idx = np.where(fold_id != fold)[0]

    train_idx, val_idx = make_inner_split(
        outer_train_idx, fold
    )

    # ----------------------------------------------------------
    # Check diplotype leakage
    # ----------------------------------------------------------
    train_groups = set(groups[train_idx])
    val_groups = set(groups[val_idx])
    test_groups = set(groups[test_idx])

    assert train_groups.isdisjoint(val_groups)
    assert train_groups.isdisjoint(test_groups)
    assert val_groups.isdisjoint(test_groups)

    print(
        f'Train      : {len(train_idx)} samples / '
        f'{len(train_groups)} diplotypes'
    )
    print(
        f'Validation : {len(val_idx)} samples / '
        f'{len(val_groups)} diplotypes'
    )
    print(
        f'Outer Test : {len(test_idx)} samples / '
        f'{len(test_groups)} diplotypes'
    )

    print(
        'Train class counts:',
        data.iloc[train_idx]['Class'].value_counts().to_dict()
    )
    print(
        'Val class counts  :',
        data.iloc[val_idx]['Class'].value_counts().to_dict()
    )
    print(
        'Test class counts :',
        data.iloc[test_idx]['Class'].value_counts().to_dict()
    )

    # ----------------------------------------------------------
    # DataLoader
    # ----------------------------------------------------------
    train_loader = DataLoader(
        Subset(dataset, train_idx),
        batch_size=BATCH_SIZE,
        shuffle=True,
        collate_fn=collate_fn
    )

    val_loader = DataLoader(
        Subset(dataset, val_idx),
        batch_size=BATCH_SIZE,
        shuffle=False,
        collate_fn=collate_fn
    )

    test_loader = DataLoader(
        Subset(dataset, test_idx),
        batch_size=BATCH_SIZE,
        shuffle=False,
        collate_fn=collate_fn
    )

    # ----------------------------------------------------------
    # Model initialization
    # ----------------------------------------------------------
    seed_everything(SEED + fold)

    model = MultimodalModel().to(device)

    # Includes CNN, GCN, and fusion/classification layers
    optimizer = torch.optim.Adam([
        {'params': model.cnn_model.parameters(), 'lr': cnn_lr},
        {'params': model.gcn_model.parameters(), 'lr': gcn_lr},
        {'params': model.fc1.parameters(),       'lr': fc_lr},
        {'params': model.fc2.parameters(),       'lr': fc_lr},
    ])

    # Best model is selected using validation Macro-F1
    best_val_f1 = -np.inf
    best_epoch = None
    patience_counter = 0

    best_path = (
        f'Multimodal_diplotype_grouped_fold_{fold}.pth'
    )

    history = []

    # ==========================================================
    # Training
    # ==========================================================
    for epoch in range(EPOCHS):

        # ------------------------------------------------------
        # Train
        # ------------------------------------------------------
        model.train()

        total_train_loss = 0.0
        total_train = 0

        train_true = []
        train_pred = []

        for xb, gb, yb in train_loader:
            xb = xb.to(device)
            gb = gb.to(device)
            yb = yb.to(device)

            optimizer.zero_grad()

            logits = model(xb, gb)
            loss = criterion(logits, yb)

            loss.backward()
            optimizer.step()

            total_train_loss += (
                loss.item() * yb.size(0)
            )
            total_train += yb.size(0)

            pred = logits.argmax(dim=1)

            train_true.extend(
                yb.detach().cpu().numpy()
            )
            train_pred.extend(
                pred.detach().cpu().numpy()
            )

        train_loss = (
            total_train_loss / total_train
        )

        train_true = np.array(train_true)
        train_pred = np.array(train_pred)

        train_acc = accuracy_score(
            train_true,
            train_pred
        )

        # ------------------------------------------------------
        # Validation
        # ------------------------------------------------------
        model.eval()

        total_val_loss = 0.0
        total_val = 0

        val_true = []
        val_probs = []

        with torch.no_grad():
            for xb, gb, yb in val_loader:
                xb = xb.to(device)
                gb = gb.to(device)
                yb = yb.to(device)

                logits = model(xb, gb)
                loss = criterion(logits, yb)

                total_val_loss += (
                    loss.item() * yb.size(0)
                )
                total_val += yb.size(0)

                probs = F.softmax(
                    logits,
                    dim=1
                )

                val_true.append(
                    yb.cpu().numpy()
                )
                val_probs.append(
                    probs.cpu().numpy()
                )

        val_loss = (
            total_val_loss / total_val
        )

        val_true = np.concatenate(
            val_true
        )

        val_probs = np.vstack(
            val_probs
        )

        val_pred = val_probs.argmax(
            axis=1
        )

        val_acc = accuracy_score(
            val_true,
            val_pred
        )

        val_precision = precision_score(
            val_true,
            val_pred,
            average='macro',
            zero_division=0
        )

        val_recall = recall_score(
            val_true,
            val_pred,
            average='macro',
            zero_division=0
        )

        val_f1 = f1_score(
            val_true,
            val_pred,
            average='macro',
            zero_division=0
        )

        # ------------------------------------------------------
        # Save training history
        # ------------------------------------------------------
        history.append({
            'fold': fold,
            'epoch': epoch + 1,
            'train_loss': train_loss,
            'train_accuracy': train_acc,
            'val_loss': val_loss,
            'val_accuracy': val_acc,
            'val_precision_macro': val_precision,
            'val_recall_macro': val_recall,
            'val_f1_macro': val_f1
        })

        # ------------------------------------------------------
        # Print epoch result
        # ------------------------------------------------------
        print(
            f'Epoch [{epoch + 1:02d}/{EPOCHS}] | '
            f'Train Loss {train_loss:.4f} '
            f'Acc {train_acc:.4f} | '
            f'Val Loss {val_loss:.4f} '
            f'Acc {val_acc:.4f} | '
            f'Val Macro-Precision {val_precision:.4f} '
            f'Macro-Recall {val_recall:.4f} '
            f'Macro-F1 {val_f1:.4f}'
        )

        # ------------------------------------------------------
        # Early stopping:
        # Best checkpoint by Validation Macro-F1
        # ------------------------------------------------------
        if val_f1 > best_val_f1 + 1e-8:
            best_val_f1 = val_f1
            best_epoch = epoch + 1
            patience_counter = 0

            torch.save(
                model.state_dict(),
                best_path
            )

        else:
            patience_counter += 1

            if patience_counter >= PATIENCE:
                print('Early stopping applied.')
                break

    # ----------------------------------------------------------
    # Save training history
    # ----------------------------------------------------------
    pd.DataFrame(history).to_csv(
        f'multimodal_training_history_fold_{fold}.csv',
        index=False
    )

    # ==========================================================
    # Reload BEST checkpoint
    # ==========================================================
    model.load_state_dict(
        torch.load(
            best_path,
            map_location=device
        )
    )

    model.eval()

    # ----------------------------------------------------------
    # Validation performance of BEST model
    # ----------------------------------------------------------
    best_val_true = []
    best_val_probs = []

    with torch.no_grad():
        for xb, gb, yb in val_loader:
            xb = xb.to(device)
            gb = gb.to(device)

            logits = model(xb, gb)

            probs = F.softmax(
                logits,
                dim=1
            )

            best_val_true.append(
                yb.numpy()
            )
            best_val_probs.append(
                probs.cpu().numpy()
            )

    best_val_true = np.concatenate(
        best_val_true
    )

    best_val_probs = np.vstack(
        best_val_probs
    )

    best_val_metrics = compute_metrics(
        best_val_true,
        best_val_probs
    )

    # ----------------------------------------------------------
    # Outer test evaluation
    # ----------------------------------------------------------
    test_true = []
    test_probs = []

    with torch.no_grad():
        for xb, gb, yb in test_loader:
            xb = xb.to(device)
            gb = gb.to(device)

            logits = model(xb, gb)

            probs = F.softmax(
                logits,
                dim=1
            )

            test_true.append(
                yb.numpy()
            )
            test_probs.append(
                probs.cpu().numpy()
            )

    yt = np.concatenate(
        test_true
    )

    pr = np.vstack(
        test_probs
    )

    test_metrics = compute_metrics(
        yt,
        pr
    )

    # ----------------------------------------------------------
    # Fold result
    # ----------------------------------------------------------
    print(f'Best epoch: {best_epoch}')

    print(
        f'Best Val Macro-F1: '
        f'{best_val_metrics["f1_macro"]:.4f}'
    )

    print(
        '*** OUTER TEST — UNSEEN DIPLOTYPES ***'
    )

    print(
        f'Acc {test_metrics["accuracy"]:.4f} | '
        f'Macro-Precision '
        f'{test_metrics["precision_macro"]:.4f} | '
        f'Macro-Recall '
        f'{test_metrics["recall_macro"]:.4f} | '
        f'Macro-F1 '
        f'{test_metrics["f1_macro"]:.4f} | '
        f'Macro-AUROC '
        f'{test_metrics["auroc_macro_ovr"]:.4f} | '
        f'Macro-AUPR '
        f'{test_metrics["aupr_macro_ovr"]:.4f}'
    )

    # ----------------------------------------------------------
    # Store fold metrics
    # ----------------------------------------------------------
    fold_row = {
        'fold': fold,
        'best_epoch': best_epoch,

        'n_train': len(train_idx),
        'n_val': len(val_idx),
        'n_test': len(test_idx),

        'n_train_groups': len(train_groups),
        'n_val_groups': len(val_groups),
        'n_test_groups': len(test_groups),

        'best_val_f1': (
            best_val_metrics['f1_macro']
        ),

        **test_metrics
    }

    fold_rows.append(
        fold_row
    )

    # ----------------------------------------------------------
    # Store OOF predictions
    # ----------------------------------------------------------
    oof_probs[test_idx] = pr

    oof_pred[test_idx] = pr.argmax(
        axis=1
    )

    del model, optimizer

    if torch.cuda.is_available():
        torch.cuda.empty_cache()


# ==============================================================
# Fold-level results
# ==============================================================
fold_results = pd.DataFrame(
    fold_rows
).sort_values('fold')

fold_results.to_csv(
    'multimodal_diplotype_grouped_5fold_metrics.csv',
    index=False
)

print('\n' + '=' * 90)
print('FOLD RESULTS')
print('=' * 90)

display(fold_results)


# ==============================================================
# Mean ± SD across outer folds
# ==============================================================
metric_cols = [
    'accuracy',
    'precision_macro',
    'recall_macro',
    'f1_macro',
    'f1_weighted',
    'auroc_macro_ovr',
    'aupr_macro_ovr',
    'nll'
]

summary = pd.DataFrame({
    'mean': fold_results[
        metric_cols
    ].mean(),

    'sd': fold_results[
        metric_cols
    ].std(ddof=1)
})

print('\n' + '=' * 90)
print('MEAN ± SD ACROSS OUTER FOLDS')
print('=' * 90)

display(summary)


# ==============================================================
# Pooled OOF performance
# ==============================================================
pooled = compute_metrics(
    y.numpy(),
    oof_probs
)

print('\n' + '=' * 90)
print('POOLED OUT-OF-FOLD PERFORMANCE')
print('=' * 90)

print(pooled)


# ==============================================================
# Save OOF predictions
# ==============================================================
oof = data.copy()

oof['fold'] = fold_id
oof['y_true'] = y.numpy()
oof['y_pred'] = oof_pred

for c, name in class_mapping.items():
    oof[f'prob_{name}'] = (
        oof_probs[:, c]
    )

oof.to_csv(
    'multimodal_diplotype_grouped_5fold_oof_predictions.csv',
    index=False
)

print(
    '\nSaved fold metrics, '
    'training histories, and OOF predictions.'
)
```
